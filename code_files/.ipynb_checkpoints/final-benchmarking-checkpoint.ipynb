```python
# =============================================================================
# CELL 0 - SHARED SETUP  (run this FIRST, once per Kaggle session)
# =============================================================================
# Defines the loaders, parsers and scoring used by all three experiments, so
# every number they produce is directly comparable. Nothing here loads a model.
#
# KAGGLE SETUP
#   Accelerator : GPU T4 x2      Internet : ON
#   Add Input   : your planner adapter dataset      (phaseA_planner)
#                 your 1-SLM adapter dataset        (slm1_improved)
# =============================================================================

import os, re, csv, gc, json, math, time, glob, traceback
from collections import Counter

import torch
from transformers import AutoModelForCausalLM, AutoTokenizer
from datasets import load_dataset

# -----------------------------------------------------------------------------
# 0.1  Kaggle's torchao is 0.10.0; peft wants >0.16.0 and RAISES instead of
#      returning False. We never use torchao - neutralise the check so the LoRA
#      dispatcher falls through to the standard Linear path.
#      Both modules must be patched: lora/torchao.py holds its own reference.
# -----------------------------------------------------------------------------
try:
    import peft.import_utils
    import peft.tuners.lora.torchao as _lt
    peft.import_utils.is_torchao_available = lambda: False
    _lt.is_torchao_available = lambda: False
    print("[patch] torchao availability check neutralised")
except Exception as e:
    print(f"[patch] skipped ({type(e).__name__})")
from peft import PeftModel

OUT_DIR = "/kaggle/working"
os.makedirs(OUT_DIR, exist_ok=True)


# -----------------------------------------------------------------------------
# 0.2  Prompts - frozen. Any change invalidates comparison with earlier runs.
# -----------------------------------------------------------------------------
PROMPT_PLANNER = ("Instruction: Provide a semantic plan without math.\n"
                  "Question: {question}\n"
                  "Plan:\n")
PROMPT_EXECUTOR = ("Solve the following math problem. A step-by-step plan is provided; follow it.\n"
                   "Reason step by step, and put your final answer within \\boxed{}.\n\n"
                   "Problem: {question}\n\n"
                   "Plan:\n{plan}\n\n"
                   "Solution:")
PROMPT_BASELINE = ("Solve the following math problem.\n"
                   "Reason step by step, and put your final answer within \\boxed{}.\n\n"
                   "Problem: {question}\n\n"
                   "Solution:")
PROMPT_1SLM = ("Solve the problem. Show only the equations, then #### answer.\n"
               "Question: {question}\n"
               "Solution:\n")

assert PROMPT_PLANNER.endswith("Plan:\n") and PROMPT_1SLM.endswith("Solution:\n")
assert "{plan}" in PROMPT_EXECUTOR and "{plan}" not in PROMPT_BASELINE

EXEC_STOPS = ("\nProblem:", "\nQuestion:", "\nPlan:", "\nSolution:")
PLAN_STOPS = ("\nQuestion:", "\nInstruction:", "Instruction:")
OPS = {"ADD", "SUBTRACT", "MULTIPLY", "DIVIDE", "CONVERT", "RATIO", "COMPARE", "COUNT"}
MAX_STEPS = 9
FALLBACK_PLAN = ("Step 1 [COUNT]: Identify the quantities given in the problem.\n"
                 "Step 2 [COMPARE]: Determine the relationship between them.\n"
                 "Step 3 [ADD]: Compute the requested quantity.")


# -----------------------------------------------------------------------------
# 0.3  Adapter discovery
# -----------------------------------------------------------------------------
def find_adapter(*hints):
    """Locate an adapter dir under /kaggle/input, preferring one matching hints."""
    hits = glob.glob("/kaggle/input/**/adapter_config.json", recursive=True)
    if not hits:
        raise FileNotFoundError("No adapter under /kaggle/input/. Add it as a Dataset.")
    for h in hits:
        low = h.lower()
        if any(x.lower() in low for x in hints):
            return os.path.dirname(h)
    return os.path.dirname(hits[0])


# -----------------------------------------------------------------------------
# 0.4  Model loading - fp16 everywhere. Never mix precisions across a table.
# -----------------------------------------------------------------------------
KNOWN_EOS = ["<|endoftext|>", "<|end_of_text|>", "<|eot_id|>", "<eos>", "</s>",
             "<|im_end|>"]


def prep_tok(tok):
    if tok.pad_token_id is None:
        tok.pad_token = tok.eos_token if tok.eos_token else "<|pad|>"
    tok.padding_side = "left"          # required for batched decoder-only gen
    return tok


def stop_ids(tok):
    ids = set()
    if tok.eos_token_id is not None:
        ids.add(tok.eos_token_id)
    for t in KNOWN_EOS:
        i = tok.convert_tokens_to_ids(t)
        if isinstance(i, int) and i >= 0 and i != tok.unk_token_id:
            ids.add(i)
    ids.discard(None)
    assert ids, "no usable EOS - generation would never stop"
    return sorted(ids)


def load_fp16(model_id):
    m = AutoModelForCausalLM.from_pretrained(
        model_id, device_map="auto", low_cpu_mem_usage=True,
        trust_remote_code=True, torch_dtype=torch.float16).eval()
    if getattr(m, "generation_config", None) is not None:
        m.generation_config.max_length = None
    return m


def load_plain(model_id):
    tok = prep_tok(AutoTokenizer.from_pretrained(model_id, trust_remote_code=True))
    m = load_fp16(model_id)
    print(f"[load] {model_id} | eos={tok.eos_token!r} stops={stop_ids(tok)}")
    return m, tok


def load_with_adapter(base_id, adapter_dir):
    """fp16 base + LoRA adapter. Handles a training-time vocab resize."""
    tok = prep_tok(AutoTokenizer.from_pretrained(adapter_dir, trust_remote_code=True))
    m = load_fp16(base_id)
    emb = m.get_input_embeddings().weight.shape[0]
    if len(tok) != emb:
        print(f"[vocab] tokenizer {len(tok)} vs embedding {emb} -> resizing")
        m.resize_token_embeddings(len(tok))
    m = PeftModel.from_pretrained(m, adapter_dir, is_trainable=False).eval()
    if getattr(m, "generation_config", None) is not None:
        m.generation_config.max_length = None
    print(f"[load] {base_id} + {os.path.basename(adapter_dir)} | stops={stop_ids(tok)}")
    return m, tok


# -----------------------------------------------------------------------------
# 0.5  Generation
# -----------------------------------------------------------------------------
@torch.no_grad()
def generate(model, tok, prompts, max_new, rep_pen=1.0):
    enc = tok(prompts, return_tensors="pt", padding=True,
              add_special_tokens=False).to(model.device)
    plen = enc["input_ids"].shape[1]
    out = model.generate(**enc, max_new_tokens=max_new, do_sample=False,
                         temperature=None, top_p=None, top_k=None,
                         repetition_penalty=rep_pen,
                         eos_token_id=stop_ids(tok), pad_token_id=tok.pad_token_id)
    return [t.strip() for t in tok.batch_decode(out[:, plen:], skip_special_tokens=True)]


def gen_safe(model, tok, prompts, max_new, rep_pen=1.0):
    try:
        return generate(model, tok, prompts, max_new, rep_pen)
    except torch.cuda.OutOfMemoryError:
        torch.cuda.empty_cache()
        if len(prompts) == 1:
            return [""]
        h = len(prompts) // 2
        print(f"   [oom] {len(prompts)} -> {h}+{len(prompts)-h}")
        return (gen_safe(model, tok, prompts[:h], max_new, rep_pen) +
                gen_safe(model, tok, prompts[h:], max_new, rep_pen))


def free(*objs):
    for o in objs:
        try:
            del o
        except Exception:
            pass
    gc.collect()
    torch.cuda.empty_cache()


# -----------------------------------------------------------------------------
# 0.6  Plan cleaning
# -----------------------------------------------------------------------------
THINK_RE = re.compile(r"<think>.*?</think>", re.DOTALL | re.IGNORECASE)
STEP_RE = re.compile(r"^[ \t]*Step[ \t]*(\d+)[ \t]*(?:\[[ \t]*([^\]]{1,32}?)[ \t]*\])?[ \t]*:[ \t]*(.*?)[ \t]*$",
                     re.MULTILINE | re.IGNORECASE)
STEP_SPLIT_RE = re.compile(r"[ \t]*(Step[ \t]*\d+[ \t]*(?:\[|:))", re.IGNORECASE)
ARITH_RE = re.compile(r"\d[ \t]*(?:[+*/×÷])[ \t]*\d|\d[ \t]+-[ \t]+\d|=[ \t]*\$?\d|<<[^>]*>>")


def cut(text, markers):
    c = len(text)
    for m in markers:
        p = text.find(m)
        if p != -1:
            c = min(c, p)
    return text[:c].strip()


def sanitise_plan(raw):
    t = THINK_RE.sub("", raw or "")
    t = re.sub(r"^\s*<think>", "", t, flags=re.IGNORECASE)
    t = cut(t, PLAN_STOPS)
    t = re.sub(r"^\s*Plan\s*:\s*", "", t, flags=re.IGNORECASE)
    t = STEP_SPLIT_RE.sub(r"\n\1", t)
    steps = []
    for _n, op, d in STEP_RE.findall(t):
        d = " ".join(d.split())
        if d:
            steps.append(((op or "").strip().upper(), d))
    steps = steps[:MAX_STEPS]
    if not steps:
        lines = [l.strip() for l in t.split("\n") if l.strip()][:MAX_STEPS]
        return {"plan": "\n".join(lines), "n_steps": 0, "format_ok": False,
                "operators": [], "ops_valid": False, "has_math": False}
    plan = "\n".join(f"Step {i} [{o}]: {d}" if o else f"Step {i}: {d}"
                     for i, (o, d) in enumerate(steps, 1))
    ops = [o for o, _ in steps]
    tagged = [o for o in ops if o]
    return {"plan": plan, "n_steps": len(steps),
            "format_ok": len(tagged) == len(ops) and bool(ops), "operators": ops,
            "ops_valid": bool(tagged) and all(o in OPS for o in tagged),
            "has_math": bool(ARITH_RE.search(plan))}


# -----------------------------------------------------------------------------
# 0.7  Answer extraction - IDENTICAL to every previous run
# -----------------------------------------------------------------------------
NUM_RE = re.compile(r"[-+]?\d[\d,]*(?:\.\d+)?|[-+]?\.\d+")
CALC_RE = re.compile(r"<<[^>]*>>")
BOXED_RE = re.compile(r"\\boxed\s*\{([^{}]*)\}")
HASH_RE = re.compile(r"####\s*([^\n]*)")
EQ_RE = re.compile(r"=\s*\$?\s*([-+]?\d[\d,]*(?:\.\d+)?)")
PHRASE_RE = re.compile(r"(?:final\s+answer|answer|result|therefore|so\s+the\s+answer)"
                       r"(?:\s*(?:\b(?:is|are|was|were|be|equals)\b|[:=]))*"
                       r"\s*\$?\s*([-+]?\d[\d,]*(?:\.\d+)?)", re.IGNORECASE)
RAMBLE = ("however", "wait", "but ", "but,", "let's correct", "let me correct",
          "there seems to be", "there is a mistake", "actually,", "on second thought",
          "hold on", "recalculat", "re-calculat", "let's re", "let me re")


def to_f(t):
    if t is None:
        return None
    c = str(t).strip().replace(",", "").replace("$", "").replace("%", "").replace("\\", "").rstrip(".")
    try:
        v = float(c)
    except ValueError:
        return None
    return v if math.isfinite(v) else None


def first_num(s):
    m = NUM_RE.search(s or "")
    return to_f(m.group(0)) if m else None


def ramble_cut(t):
    low, c = t.lower(), len(t)
    for m in RAMBLE:
        p = low.find(m)
        if p != -1:
            c = min(c, p)
    return c


def modal(t):
    vals = [v for v in (to_f(m.group(0)) for m in NUM_RE.finditer(t)) if v is not None]
    if not vals:
        return None
    cnt = Counter(vals)
    best, bc = None, 1
    for v in vals:
        if cnt[v] > bc:
            best, bc = v, cnt[v]
    return (best, bc) if best is not None else None


def extract_answer(text):
    """Returns (value, strategy). Strips <think> blocks first - reasoning-distilled
    models emit them and the answer is always after the closing tag."""
    if not text:
        return None, "empty"
    c = THINK_RE.sub("", text)
    if "<think>" in c.lower() and "</think>" not in c.lower():
        # unterminated think block: the model never finished reasoning
        c = re.sub(r"<think>.*", "", c, flags=re.DOTALL | re.IGNORECASE)
    c = CALC_RE.sub("", c)
    h = HASH_RE.findall(c)
    if h and first_num(h[-1]) is not None:
        return first_num(h[-1]), "hash"
    b = BOXED_RE.findall(c)
    if b and first_num(b[-1]) is not None:
        return first_num(b[-1]), "boxed"
    pre = c[:ramble_cut(c)]
    p = PHRASE_RE.findall(pre)
    if p and to_f(p[-1]) is not None:
        return to_f(p[-1]), "phrase_preramble"
    e = EQ_RE.findall(pre)
    if e and to_f(e[-1]) is not None:
        return to_f(e[-1]), "equals_preramble"
    m = modal(pre)
    if m and m[1] >= 2:
        return m[0], "modal_preramble"
    n = NUM_RE.findall(pre)
    if n and to_f(n[-1]) is not None:
        return to_f(n[-1]), "last_number_preramble"
    n = NUM_RE.findall(c)
    if n and to_f(n[-1]) is not None:
        return to_f(n[-1]), "last_number"
    return None, "none"


def correct(pred, gold):
    return (pred is not None and gold is not None
            and math.isclose(pred, gold, rel_tol=1e-4, abs_tol=1e-4))


def mcnemar(b, c):
    from math import comb
    n = b + c
    if n == 0:
        return 1.0
    return min(2 * sum(comb(n, i) for i in range(max(b, c), n + 1)) * 0.5 ** n, 1.0)


def wilson_ci(k, n):
    if n == 0:
        return (0.0, 0.0)
    p = k / n
    se = math.sqrt(p * (1 - p) / n)
    return (p - 1.96 * se, p + 1.96 * se)


# -----------------------------------------------------------------------------
# 0.8  Dataset loaders with column auto-detection
# -----------------------------------------------------------------------------
def load_gsm8k(split="test"):
    ds = load_dataset("openai/gsm8k", "main", split=split)
    qs = [" ".join(r["question"].split()) for r in ds]
    gs = [first_num(CALC_RE.sub("", r["answer"].split("####")[-1])) for r in ds]
    return qs, gs, "GSM8K"


def _pick(row, names):
    for n in names:
        if n in row and row[n] not in (None, ""):
            return row[n]
    return None


def load_svamp():
    """SVAMP: elementary word problems. Body + Question must be concatenated."""
    last = None
    for repo in ["ChilleD/SVAMP", "arkilpatel/SVAMP"]:
        for split in ["test", "train"]:
            try:
                ds = load_dataset(repo, split=split)
                qs, gs = [], []
                for r in ds:
                    body = _pick(r, ["Body", "body"]) or ""
                    ques = _pick(r, ["Question", "question"]) or ""
                    ans = _pick(r, ["Answer", "answer", "final_ans", "Equation"])
                    q = " ".join((str(body) + " " + str(ques)).split())
                    g = to_f(ans) if not isinstance(ans, str) else first_num(str(ans))
                    if q and g is not None:
                        qs.append(q)
                        gs.append(g)
                if qs:
                    print(f"[data] SVAMP from {repo}/{split}: {len(qs)} problems")
                    return qs, gs, "SVAMP"
            except Exception as e:
                last = e
    raise RuntimeError(f"could not load SVAMP: {last}")


def load_multiarith():
    last = None
    for repo in ["ChilleD/MultiArith"]:
        for split in ["test", "train"]:
            try:
                ds = load_dataset(repo, split=split)
                qs, gs = [], []
                for r in ds:
                    q = _pick(r, ["question", "Question"])
                    a = _pick(r, ["final_ans", "answer", "Answer"])
                    g = to_f(a) if not isinstance(a, str) else first_num(str(a))
                    if q and g is not None:
                        qs.append(" ".join(str(q).split()))
                        gs.append(g)
                if qs:
                    print(f"[data] MultiArith from {repo}/{split}: {len(qs)} problems")
                    return qs, gs, "MultiArith"
            except Exception as e:
                last = e
    raise RuntimeError(f"could not load MultiArith: {last}")


# -----------------------------------------------------------------------------
# 0.9  CSV helpers
# -----------------------------------------------------------------------------
def csv_init(path, fields):
    if not os.path.exists(path) or os.path.getsize(path) == 0:
        with open(path, "w", newline="", encoding="utf-8") as f:
            csv.DictWriter(f, fieldnames=fields).writeheader()
        return set()
    return {int(r["idx"]) for r in csv.DictReader(open(path, newline=""))
            if r.get("idx", "").strip().isdigit()}


def csv_append(path, fields, rows):
    with open(path, "a", newline="", encoding="utf-8") as f:
        w = csv.DictWriter(f, fieldnames=fields, extrasaction="ignore")
        for r in rows:
            w.writerow({k: r.get(k, "") for k in fields})
        f.flush()
        os.fsync(f.fileno())


print("\n[common] ready. Prompts frozen, scoring identical to previous runs.")
print("[common] fp16 everywhere - do not mix with 4-bit numbers.")
```

```python
# =============================================================================
# REPLACEMENT 1 - paste right after CELL0_common to override the broken free()
# =============================================================================
import os, gc, torch
 
# Reduces fragmentation; helps when a big contiguous alloc (the resize) follows
# many small ones.
os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"
 
 
def sweep():
    """Collect and release cached VRAM. Call AFTER you `del` your references."""
    gc.collect()
    torch.cuda.empty_cache()
    torch.cuda.synchronize()
 
 
def vram(tag=""):
    parts = []
    for i in range(torch.cuda.device_count()):
        a = torch.cuda.memory_allocated(i) / 1024**3
        t = torch.cuda.get_device_properties(i).total_memory / 1024**3
        parts.append(f"gpu{i} {a:.1f}/{t:.0f}GB")
    print(f"[vram] {tag:<24} " + "  ".join(parts))
 
 
def free(*objs):
    """Kept for compatibility, but it CANNOT free the caller's references.
    Always do `del model, tok` yourself, then call sweep()."""
    sweep()
 
 
# Spread big models over both T4s instead of filling gpu0 first.
MAX_MEM = {i: "13GiB" for i in range(torch.cuda.device_count())}
MAX_MEM["cpu"] = "24GiB"
 
 
def load_fp16(model_id):                       # overrides CELL0's version
    m = AutoModelForCausalLM.from_pretrained(
        model_id, device_map="auto", max_memory=MAX_MEM,
        low_cpu_mem_usage=True, trust_remote_code=True,
        torch_dtype=torch.float16).eval()
    if getattr(m, "generation_config", None) is not None:
        m.generation_config.max_length = None
    return m
 
 
def load_with_adapter(base_id, adapter_dir):   # overrides CELL0's version
    """Resize BEFORE the model is large in VRAM is not possible with device_map,
    so instead we only resize when genuinely needed and report it."""
    tok = prep_tok(AutoTokenizer.from_pretrained(adapter_dir, trust_remote_code=True))
    m = load_fp16(base_id)
    emb = m.get_input_embeddings().weight.shape[0]
    if len(tok) > emb:
        print(f"[vocab] tokenizer {len(tok)} > embedding {emb} -> resizing "
              f"(needs ~{(len(tok)-emb)*m.config.hidden_size*2/1024**2:.0f} MB + a full copy)")
        m.resize_token_embeddings(len(tok))
    elif len(tok) < emb:
        # Harmless: extra unused embedding rows. Resizing DOWN would break the
        # adapter, so leave it.
        print(f"[vocab] tokenizer {len(tok)} < embedding {emb} - leaving as is")
    m = PeftModel.from_pretrained(m, adapter_dir, is_trainable=False).eval()
    if getattr(m, "generation_config", None) is not None:
        m.generation_config.max_length = None
    return m, tok
 
 
vram("after CELL0")
print("[patch] free() fixed -> use `del model, tok; sweep()` from now on")
```

```python
# =============================================================================
# EXPERIMENT 1 - MATCHED LATENCY FOR ALL CONFIGURATIONS
# =============================================================================
# Run CELL0_common first.
#
# WHY RE-MEASURE EVERYTHING, not just the two blanks:
#   your 26.69s and 16.05s were measured at 4-bit on Colab; the 4.11s was fp16
#   on Kaggle. 4-bit inference carries dequantisation overhead, so the old
#   figures are inflated and the 6.5x speedup is an upper bound. Filling only
#   the gaps would leave a table with three different measurement conditions in
#   one column. This re-measures all five on one machine, fp16, sequential
#   batch=1, so the column is internally consistent.
#
# Sequential batch=1 is deliberate: it is the per-request latency a deployment
# actually sees. Batched throughput is a different quantity and must not be
# mixed into the same column.
#
# TIME: ~45-60 min for N_LAT=30. Raise to 50 if you have room.
# =============================================================================

N_LAT = 30                  # problems per configuration, sequential
PLANNER_BASE = "Qwen/Qwen3-4B"
EXEC_MATH = "Qwen/Qwen2.5-Math-1.5B"
EXEC_GEN = "Qwen/Qwen2.5-1.5B"
SLM1_BASE = "Qwen/Qwen2.5-Math-1.5B"
PLANNER_MAX_NEW, EXEC_MAX_NEW, SLM1_MAX_NEW = 192, 640, 128
REP_PEN = 1.15

PLANNER_ADAPTER = find_adapter("phaseA", "planner")
SLM1_ADAPTER = find_adapter("slm1", "improved", "compress")
print(f"[cfg] planner adapter : {PLANNER_ADAPTER}")
print(f"[cfg] 1-SLM adapter   : {SLM1_ADAPTER}")

QS, GS, _ = load_gsm8k("test")
QS, GS = QS[:N_LAT], GS[:N_LAT]
LAT = {}


def timed(fn, n=N_LAT):
    """Median is reported alongside mean: one slow outlier should not set the
    headline number, and generation length varies a lot per problem."""
    ts = []
    for i in range(n):
        t0 = time.time()
        fn(i)
        ts.append(time.time() - t0)
    ts_sorted = sorted(ts)
    return sum(ts) / len(ts), ts_sorted[len(ts) // 2]


# -----------------------------------------------------------------------------
# 1.1  Planner once, reused by both 2-SLM configurations
# -----------------------------------------------------------------------------
print("\n[1/3] planner + both executors (2-SLM end-to-end, and Phase B alone)")
planner, ptok = load_with_adapter(PLANNER_BASE, PLANNER_ADAPTER)

# Pre-time the planner alone so the 2-SLM split (plan vs execute) is visible.
plan_cache = {}


def _plan(i):
    raw = gen_safe(planner, ptok, [PROMPT_PLANNER.replace("{question}", QS[i])],
                   PLANNER_MAX_NEW)[0]
    plan_cache[i] = sanitise_plan(raw)["plan"].strip() or FALLBACK_PLAN


m, md = timed(_plan)
LAT["Phase A (planner) alone"] = (m, md)
print(f"   planner: mean {m:.2f}s median {md:.2f}s")

for exec_id, label in [(EXEC_MATH, "Qwen-Math"), (EXEC_GEN, "Qwen general")]:
    ex, etok = load_plain(exec_id)

    def _guided(i):
        gen_safe(ex, etok, [PROMPT_EXECUTOR.replace("{question}", QS[i])
                            .replace("{plan}", plan_cache[i])], EXEC_MAX_NEW, REP_PEN)

    def _alone(i):
        gen_safe(ex, etok, [PROMPT_BASELINE.replace("{question}", QS[i])],
                 EXEC_MAX_NEW, REP_PEN)

    m_g, md_g = timed(_guided)
    m_a, md_a = timed(_alone)
    # 2-SLM end-to-end = planner + executor, since they run sequentially
    LAT[f"2-SLM guided ({label})"] = (LAT["Phase A (planner) alone"][0] + m_g,
                                      LAT["Phase A (planner) alone"][1] + md_g)
    LAT[f"Phase B alone ({label})"] = (m_a, md_a)
    print(f"   {label}: guided-exec {m_g:.2f}s | alone {m_a:.2f}s | "
          f"2-SLM total {LAT[f'2-SLM guided ({label})'][0]:.2f}s")
    free(ex)

free(planner)

# -----------------------------------------------------------------------------
# 1.2  1-SLM compressed
# -----------------------------------------------------------------------------
print("\n[2/3] 1-SLM compressed")
s1, s1tok = load_with_adapter(SLM1_BASE, SLM1_ADAPTER)


def _slm1(i):
    gen_safe(s1, s1tok, [PROMPT_1SLM.replace("{question}", QS[i])], SLM1_MAX_NEW)


m, md = timed(_slm1)
LAT["1-SLM compressed"] = (m, md)
print(f"   1-SLM: mean {m:.2f}s median {md:.2f}s")
free(s1)

# -----------------------------------------------------------------------------
# 1.3  Table
# -----------------------------------------------------------------------------
ACC = {  # fp16, n=1319, from your completed runs
    "2-SLM guided (Qwen-Math)": 0.7160,
    "Phase B alone (Qwen-Math)": 0.7207,
    "2-SLM guided (Qwen general)": 0.5600,
    "Phase B alone (Qwen general)": 0.4550,
    "1-SLM compressed": 0.6975,
}
ref = LAT["2-SLM guided (Qwen-Math)"][0]

print("\n" + "=" * 88)
print(f"LATENCY, ALL fp16, SEQUENTIAL batch=1, SAME MACHINE  (n={N_LAT} per config)")
print("=" * 88)
print(f"{'configuration':<32}{'acc':>8}{'mean':>10}{'median':>10}{'speedup':>10}")
print("-" * 88)
for k in ["2-SLM guided (Qwen-Math)", "Phase B alone (Qwen-Math)",
          "2-SLM guided (Qwen general)", "Phase B alone (Qwen general)",
          "1-SLM compressed", "Phase A (planner) alone"]:
    if k not in LAT:
        continue
    mean, med = LAT[k]
    a = ACC.get(k)
    astr = f"{a:.1%}" if a else "-"
    print(f"{k:<32}{astr:>8}{mean:>9.2f}s{med:>9.2f}s{ref/mean:>9.1f}x")
print("-" * 88)

json.dump({k: {"mean_s": v[0], "median_s": v[1]} for k, v in LAT.items()},
          open(f"{OUT_DIR}/latency_matched.json", "w"), indent=2)
print(f"\nsaved -> {OUT_DIR}/latency_matched.json")
print(f"""
FOR THE PAPER
  Report THESE latencies, not the Colab 4-bit ones. State the measurement
  condition explicitly: "fp16, sequential batch=1, single NVIDIA T4, n={N_LAT}".
  The speedup will likely be smaller than the earlier 6.5x, because the old
  2-SLM figure was inflated by 4-bit dequantisation overhead. A smaller,
  honestly-measured number is worth more than a large unreproducible one.
""")
```

```python
# =============================================================================
# PART A - GENERATE PLANS FIRST, planner alone in VRAM
# =============================================================================
# The planner (Qwen3-4B fp16, ~8 GB) is by far the largest model here. Running
# it before any executor guarantees it has the whole card. Plans are cached to
# disk, so PART B never needs it again even if the session dies.
# =============================================================================
PLANNER_BASE = "Qwen/Qwen3-4B"
PLANNER_ADAPTER = find_adapter("phaseA", "planner")
N_PROBLEMS = 1319
PLAN_CSV = f"{OUT_DIR}/plans_gsm8k_test.csv"
PLAN_BS = 16
 
QS, GS, DSNAME = load_gsm8k("test")
N = min(N_PROBLEMS, len(QS))
 
PLANS = [None] * N
# reuse from any CSV that already has a plan column
for pat in [PLAN_CSV, "/kaggle/working/twoslm_full_1319.csv", "/kaggle/input/**/*.csv"]:
    for h in sorted(glob.glob(pat, recursive=True)):
        try:
            rdr = csv.DictReader(open(h, newline="", encoding="utf-8"))
            if "plan" not in (rdr.fieldnames or []):
                continue
            got = 0
            for r in rdr:
                if r.get("idx", "").strip().isdigit() and not r.get("error"):
                    i = int(r["idx"])
                    if 0 <= i < N and r.get("plan", "").strip():
                        PLANS[i] = r["plan"]
                        got += 1
            if got:
                print(f"[plans] reused {got}/{N} from {h}")
                break
        except Exception:
            continue
    if any(p is not None for p in PLANS):
        break
 
missing = [i for i in range(N) if PLANS[i] is None]
if missing:
    print(f"[plans] generating {len(missing)} with the planner alone in VRAM")
    vram("before planner")
    planner, ptok = load_with_adapter(PLANNER_BASE, PLANNER_ADAPTER)
    vram("planner loaded")
    pf = ["idx", "plan"]
    csv_init(PLAN_CSV, pf)
    t0 = time.time()
    for s in range(0, len(missing), PLAN_BS):
        ks = missing[s:s + PLAN_BS]
        raws = gen_safe(planner, ptok,
                        [PROMPT_PLANNER.replace("{question}", QS[k]) for k in ks], 192)
        rows = []
        for k, raw in zip(ks, raws):
            PLANS[k] = sanitise_plan(raw)["plan"]
            rows.append({"idx": k, "plan": PLANS[k]})
        csv_append(PLAN_CSV, pf, rows)
        if (s // PLAN_BS) % 10 == 0:
            el = time.time() - t0
            dn = s + len(ks)
            print(f"   plans {dn}/{len(missing)} | ETA "
                  f"{(len(missing)-dn)*el/max(dn,1)/60:.0f}m", flush=True)
    # THE FIX: caller dels, then sweeps.
    del planner, ptok
    sweep()
    vram("planner released")
 
PLANS = [p.strip() if p and p.strip() else FALLBACK_PLAN for p in PLANS]
print(f"[plans] ready: {len(PLANS)}  (cached at {PLAN_CSV})")
```

```python
# =============================================================================
# PART B - PROBE + EVALUATE  (run after PART A, same session is fine now)
# =============================================================================
CANDIDATES = [
    ("Qwen/Qwen2.5-0.5B", "qwen2.5-0.5B", 24),
    ("Qwen/Qwen2.5-3B",   "qwen2.5-3B",    8),
]
PROBE_N = 60
PROBE_MIN, PROBE_MAX = 0.15, 0.68
MAX_NEW, REP_PEN = 640, 1.15
 
print("\n" + "=" * 78)
print(f"PRE-FLIGHT PROBE (baseline only, n={PROBE_N}, band {PROBE_MIN:.0%}-{PROBE_MAX:.0%})")
print("=" * 78)
SELECTED = []
for model_id, tag, bs in CANDIDATES:
    m = tk = None
    try:
        m, tk = load_plain(model_id)
        okc = 0
        for s in range(0, PROBE_N, bs):
            outs = gen_safe(m, tk, [PROMPT_BASELINE.replace("{question}", q)
                                    for q in QS[s:s + bs]], MAX_NEW, REP_PEN)
            for g, raw in zip(GS[s:s + bs], outs):
                p, _ = extract_answer(cut(raw, EXEC_STOPS))
                okc += int(correct(p, g))
        acc = okc / PROBE_N
        verdict = ("SKIP - floor effect" if acc < PROBE_MIN else
                   "SKIP - already strong" if acc > PROBE_MAX else "RUN")
        if verdict == "RUN":
            SELECTED.append((model_id, tag, bs, acc))
        print(f"  {tag:<16} baseline {acc:>6.1%} -> {verdict}")
    except Exception as e:
        print(f"  {tag:<16} FAILED: {type(e).__name__}: {str(e)[:60]}")
    finally:
        if m is not None:
            del m
        if tk is not None:
            del tk
        sweep()
        vram(f"after probe {tag}")
 
if not SELECTED:
    raise SystemExit("Nothing in the usable band.")
print(f"proceeding with: {[s[1] for s in SELECTED]}")
 
FIELDS = ["idx", "question", "gold", "plan", "guided_raw", "guided_pred",
          "guided_strategy", "guided_correct", "baseline_raw", "baseline_pred",
          "baseline_strategy", "baseline_correct"]
RESULTS = {}
 
for model_id, tag, bs, probe_acc in SELECTED:
    out_csv = f"{OUT_DIR}/grad_{tag}.csv"
    done = csv_init(out_csv, FIELDS)
    todo = [i for i in range(N) if i not in done]
    print("\n" + "=" * 78)
    print(f"{model_id}  probe={probe_acc:.1%}  {len(todo)} to run")
    print("=" * 78)
    if not todo:
        continue
 
    model, tok = load_plain(model_id)
    vram(f"{tag} loaded")
    st, strat = Counter(), Counter()
    t0 = time.time()
    for s in range(0, len(todo), bs):
        ks = todo[s:s + bs]
        qs, pls, gs = [QS[k] for k in ks], [PLANS[k] for k in ks], [GS[k] for k in ks]
        rows = []
        try:
            gt = [cut(a, EXEC_STOPS) for a in gen_safe(
                model, tok, [PROMPT_EXECUTOR.replace("{question}", q).replace("{plan}", p)
                             for q, p in zip(qs, pls)], MAX_NEW, REP_PEN)]
            bt = [cut(a, EXEC_STOPS) for a in gen_safe(
                model, tok, [PROMPT_BASELINE.replace("{question}", q) for q in qs],
                MAX_NEW, REP_PEN)]
            for k, q, p, g, ga, ba in zip(ks, qs, pls, gs, gt, bt):
                gp, gsx = extract_answer(ga)
                bp, bsx = extract_answer(ba)
                gok, bok = correct(gp, g), correct(bp, g)
                st["n"] += 1
                st["g"] += int(gok)
                st["b"] += int(bok)
                st["go"] += int(gok and not bok)
                st["bo"] += int(bok and not gok)
                strat[gsx] += 1
                rows.append({"idx": k, "question": q, "gold": g, "plan": p,
                             "guided_raw": ga, "guided_pred": gp, "guided_strategy": gsx,
                             "guided_correct": gok, "baseline_raw": ba,
                             "baseline_pred": bp, "baseline_strategy": bsx,
                             "baseline_correct": bok})
        except Exception:
            traceback.print_exc(limit=1)
            sweep()
        csv_append(out_csv, FIELDS, rows)
        if st["n"]:
            el = time.time() - t0
            print(f"[{st['n']}/{len(todo)}] guided={st['g']/st['n']:.1%} "
                  f"base={st['b']/st['n']:.1%} delta={(st['g']-st['b'])/st['n']:+.1%} "
                  f"| ETA {(len(todo)-st['n'])*el/st['n']/3600:.2f}h", flush=True)
 
    n = max(st["n"], 1)
    RESULTS[tag] = {"model": model_id, "n": st["n"], "baseline": st["b"] / n,
                    "guided": st["g"] / n, "delta": (st["g"] - st["b"]) / n,
                    "go": st["go"], "bo": st["bo"], "p": mcnemar(st["go"], st["bo"])}
    r = RESULTS[tag]
    print(f"\n[{tag}] base {r['baseline']:.2%} guided {r['guided']:.2%} "
          f"delta {r['delta']:+.2%} p={r['p']:.4f}")
    del model, tok
    sweep()
    vram(f"{tag} released")
 
# -----------------------------------------------------------------------------
# GRADIENT TABLE
# -----------------------------------------------------------------------------
rows = [("Qwen2.5-1.5B (general)", 0.4550, 0.5600, mcnemar(310, 172), 1319),
        ("Qwen2.5-Math-1.5B", 0.7207, 0.7160, None, 1271)]
for tag, r in RESULTS.items():
    rows.append((r["model"].split("/")[-1], r["baseline"], r["guided"], r["p"], r["n"]))
rows.sort(key=lambda x: x[1])
 
print("\n" + "=" * 92)
print("GUIDANCE BENEFIT vs EXECUTOR UNAIDED ABILITY")
print("=" * 92)
print(f"{'executor':<28}{'unaided':>10}{'guided':>10}{'delta':>10}{'p':>11}{'n':>7}")
print("-" * 92)
for name, b, g, p, nn in rows:
    ps = f"{p:.4f}" if p is not None else "-"
    star = " *" if (p is not None and p < 0.05) else ""
    print(f"{name:<28}{b:>9.1%}{g:>10.1%}{g-b:>+10.1%}{ps:>11}{nn:>7}{star}")
print("-" * 92)
xs = [b for _, b, _, _, _ in rows]
ys = [g - b for _, b, g, _, _ in rows]
if len(xs) >= 3:
    mx, my = sum(xs) / len(xs), sum(ys) / len(ys)
    num = sum((x - mx) * (y - my) for x, y in zip(xs, ys))
    dx = math.sqrt(sum((x - mx) ** 2 for x in xs))
    dy = math.sqrt(sum((y - my) ** 2 for y in ys))
    if dx and dy:
        print(f"\n  correlation(unaided, benefit) = {num/(dx*dy):+.3f}")
 
json.dump(RESULTS, open(f"{OUT_DIR}/capability_gradient.json", "w"), indent=2)
print(f"\nsaved -> {OUT_DIR}/capability_gradient.json")
```

```python
# =============================================================================
# EXPERIMENT 3 (v2) - OUT-OF-DISTRIBUTION: SVAMP and MultiArith
# =============================================================================
# Replaces EXP3_ood_datasets.py. Two fixes:
#
#  1. VRAM. The old version called free(model), which cannot release a caller's
#     reference, so nothing was freed. Per dataset it loaded planner (8GB) +
#     executor (3GB) + planner AGAIN (3rd load) + 1-SLM (3GB) = ~22GB on a
#     14.5GB card. Now: `del` in the caller, then sweep(), with vram()
#     checkpoints so a regression is visible immediately.
#
#  2. The planner was loaded twice per dataset - once for plans, once for
#     latency. Now loaded once, and its latency is measured while it is already
#     resident.
#
# PREREQUISITES in this session:
#     CELL0_common            (loaders, parsers, scoring)
#     Replacement 1 from PATCH_vram_fix  (sweep, vram, fixed load_* overrides)
#
# WHY THIS EXPERIMENT MATTERS MOST
#   Both your planner and your 1-SLM trained on GSM8K. "Did you measure
#   reasoning or memorised distribution?" is the first question a reviewer asks.
#   SVAMP was built specifically to catch pattern-matching. Neither model has
#   seen either dataset.
#
# TIME: ~1.5-2h for both datasets, all three configs. Resumable.
# =============================================================================

DATASETS = ["svamp", "multiarith"]     # drop one if time is short
EXEC_GEN = "Qwen/Qwen2.5-1.5B"
SLM1_BASE = "Qwen/Qwen2.5-Math-1.5B"
PLANNER_BASE = "Qwen/Qwen3-4B"
BATCH_SIZE = 16
PLANNER_MAX_NEW, EXEC_MAX_NEW, SLM1_MAX_NEW = 192, 640, 128
REP_PEN = 1.15
N_LAT = 25                              # sequential subset for latency

PLANNER_ADAPTER = find_adapter("phaseA", "planner")
SLM1_ADAPTER = find_adapter("slm1", "improved", "compress")
LOADERS = {"svamp": load_svamp, "multiarith": load_multiarith, "gsm8k": load_gsm8k}
ALL = {}

assert "sweep" in dir(), "Run Replacement 1 from PATCH_vram_fix first (defines sweep/vram)."

for dsname in DATASETS:
    try:
        QS, GS, LABEL = LOADERS[dsname]()
    except Exception as e:
        print(f"[skip] {dsname}: {type(e).__name__}: {e}")
        continue
    N = len(QS)
    print("\n" + "#" * 78)
    print(f"# {LABEL}  ({N} problems)")
    print("#" * 78)
    res = {"n": N}

    # -------------------------------------------------------------------------
    # 3.1  Plans + planner latency, planner loaded ONCE
    # -------------------------------------------------------------------------
    plan_csv = f"{OUT_DIR}/plans_{dsname}.csv"
    PLANS = [None] * N
    if os.path.exists(plan_csv):
        for r in csv.DictReader(open(plan_csv, newline="", encoding="utf-8")):
            if r.get("idx", "").strip().isdigit():
                i = int(r["idx"])
                if 0 <= i < N:
                    PLANS[i] = r.get("plan", "")
        print(f"[plans] reused {sum(p is not None for p in PLANS)}/{N}")

    missing = [i for i in range(N) if PLANS[i] is None]
    need_planner_lat = "planner_lat" not in res
    if missing or need_planner_lat:
        vram("before planner")
        planner, ptok = load_with_adapter(PLANNER_BASE, PLANNER_ADAPTER)
        vram("planner loaded")

        if missing:
            pf = ["idx", "plan", "format_ok", "ops_valid", "has_math"]
            csv_init(plan_csv, pf)
            fmt = Counter()
            t0 = time.time()
            for s in range(0, len(missing), BATCH_SIZE):
                ks = missing[s:s + BATCH_SIZE]
                raws = gen_safe(planner, ptok,
                                [PROMPT_PLANNER.replace("{question}", QS[k]) for k in ks],
                                PLANNER_MAX_NEW)
                rows = []
                for k, raw in zip(ks, raws):
                    inf = sanitise_plan(raw)
                    PLANS[k] = inf["plan"]
                    fmt["ok"] += int(inf["format_ok"])
                    fmt["ops"] += int(inf["ops_valid"])
                    fmt["math"] += int(inf["has_math"])
                    fmt["n"] += 1
                    rows.append({"idx": k, "plan": inf["plan"],
                                 "format_ok": inf["format_ok"],
                                 "ops_valid": inf["ops_valid"],
                                 "has_math": inf["has_math"]})
                csv_append(plan_csv, pf, rows)
                if (s // BATCH_SIZE) % 5 == 0:
                    dn = s + len(ks)
                    print(f"   plans {dn}/{len(missing)} | ETA "
                          f"{(len(missing)-dn)*(time.time()-t0)/max(dn,1)/60:.0f}m", flush=True)
            nf = max(fmt["n"], 1)
            # Plan quality on unseen data is itself a result: the planner only
            # ever saw GSM8K plans, so this measures format transfer.
            print(f"[plans] OOD quality: format {fmt['ok']/nf:.1%} | "
                  f"ops-in-vocab {fmt['ops']/nf:.1%} | arithmetic leak {fmt['math']/nf:.1%}")
            res["plan_format_ok"] = fmt["ok"] / nf
            res["plan_ops_valid"] = fmt["ops"] / nf
            res["plan_has_math"] = fmt["math"] / nf

        # planner latency while it is already loaded - no second load
        lp = []
        for i in range(min(N_LAT, N)):
            t1 = time.time()
            gen_safe(planner, ptok, [PROMPT_PLANNER.replace("{question}", QS[i])],
                     PLANNER_MAX_NEW)
            lp.append(time.time() - t1)
        res["planner_lat"] = sum(lp) / len(lp)
        print(f"[lat] planner {res['planner_lat']:.2f}s")

        del planner, ptok
        sweep()
        vram("planner released")

    PLANS = [p.strip() if p and p.strip() else FALLBACK_PLAN for p in PLANS]

    # -------------------------------------------------------------------------
    # 3.2  General executor: guided vs alone
    # -------------------------------------------------------------------------
    out_csv = f"{OUT_DIR}/ood_{dsname}_general.csv"
    F = ["idx", "question", "gold", "plan", "guided_raw", "guided_pred",
         "guided_strategy", "guided_correct", "baseline_raw", "baseline_pred",
         "baseline_strategy", "baseline_correct"]
    done = csv_init(out_csv, F)
    todo = [i for i in range(N) if i not in done]

    model, tok = load_plain(EXEC_GEN)
    vram("general executor loaded")
    st = Counter()
    t0 = time.time()
    for s in range(0, len(todo), BATCH_SIZE):
        ks = todo[s:s + BATCH_SIZE]
        qs, pls, gs = [QS[k] for k in ks], [PLANS[k] for k in ks], [GS[k] for k in ks]
        rows = []
        try:
            gt = [cut(a, EXEC_STOPS) for a in gen_safe(
                model, tok, [PROMPT_EXECUTOR.replace("{question}", q).replace("{plan}", p)
                             for q, p in zip(qs, pls)], EXEC_MAX_NEW, REP_PEN)]
            bt = [cut(a, EXEC_STOPS) for a in gen_safe(
                model, tok, [PROMPT_BASELINE.replace("{question}", q) for q in qs],
                EXEC_MAX_NEW, REP_PEN)]
            for k, q, p, g, ga, ba in zip(ks, qs, pls, gs, gt, bt):
                gp, gsx = extract_answer(ga)
                bp, bsx = extract_answer(ba)
                gok, bok = correct(gp, g), correct(bp, g)
                st["n"] += 1
                st["g"] += int(gok)
                st["b"] += int(bok)
                st["go"] += int(gok and not bok)
                st["bo"] += int(bok and not gok)
                rows.append({"idx": k, "question": q, "gold": g, "plan": p,
                             "guided_raw": ga, "guided_pred": gp, "guided_strategy": gsx,
                             "guided_correct": gok, "baseline_raw": ba,
                             "baseline_pred": bp, "baseline_strategy": bsx,
                             "baseline_correct": bok})
        except Exception:
            traceback.print_exc(limit=1)
            sweep()
        csv_append(out_csv, F, rows)
        if st["n"] and (s // BATCH_SIZE) % 5 == 0:
            el = time.time() - t0
            print(f"   [{st['n']}/{len(todo)}] guided={st['g']/st['n']:.1%} "
                  f"base={st['b']/st['n']:.1%} | ETA "
                  f"{(len(todo)-st['n'])*el/st['n']/60:.0f}m", flush=True)

    ng = max(st["n"], 1)
    res.update({"general_baseline": st["b"] / ng, "general_guided": st["g"] / ng,
                "general_delta": (st["g"] - st["b"]) / ng,
                "general_go": st["go"], "general_bo": st["bo"],
                "general_p": mcnemar(st["go"], st["bo"]), "general_n": st["n"]})
    print(f"[{LABEL}] general: base {res['general_baseline']:.2%} "
          f"guided {res['general_guided']:.2%} delta {res['general_delta']:+.2%} "
          f"p={res['general_p']:.4f}")

    lg, lb = [], []
    for i in range(min(N_LAT, N)):
        t1 = time.time()
        gen_safe(model, tok, [PROMPT_EXECUTOR.replace("{question}", QS[i])
                              .replace("{plan}", PLANS[i])], EXEC_MAX_NEW, REP_PEN)
        lg.append(time.time() - t1)
        t1 = time.time()
        gen_safe(model, tok, [PROMPT_BASELINE.replace("{question}", QS[i])],
                 EXEC_MAX_NEW, REP_PEN)
        lb.append(time.time() - t1)
    res["general_exec_lat"] = sum(lg) / len(lg)
    res["general_alone_lat"] = sum(lb) / len(lb)
    res["twoslm_lat"] = res.get("planner_lat", 0) + res["general_exec_lat"]
    print(f"[lat] exec-guided {res['general_exec_lat']:.2f}s | "
          f"alone {res['general_alone_lat']:.2f}s | 2-SLM total {res['twoslm_lat']:.2f}s")

    del model, tok
    sweep()
    vram("general executor released")

    # -------------------------------------------------------------------------
    # 3.3  1-SLM compressed
    # -------------------------------------------------------------------------
    out_csv = f"{OUT_DIR}/ood_{dsname}_1slm.csv"
    F2 = ["idx", "question", "gold", "raw", "pred", "strategy", "correct"]
    done = csv_init(out_csv, F2)
    todo = [i for i in range(N) if i not in done]

    s1, s1tok = load_with_adapter(SLM1_BASE, SLM1_ADAPTER)
    vram("1-SLM loaded")
    st2, strat2 = Counter(), Counter()
    for s in range(0, len(todo), BATCH_SIZE):
        ks = todo[s:s + BATCH_SIZE]
        qs, gs = [QS[k] for k in ks], [GS[k] for k in ks]
        rows = []
        try:
            outs = gen_safe(s1, s1tok,
                            [PROMPT_1SLM.replace("{question}", q) for q in qs],
                            SLM1_MAX_NEW)
            for k, q, g, raw in zip(ks, qs, gs, outs):
                p, sx = extract_answer(raw)
                c = correct(p, g)
                st2["n"] += 1
                st2["ok"] += int(c)
                strat2[sx] += 1
                rows.append({"idx": k, "question": q, "gold": g, "raw": raw,
                             "pred": p, "strategy": sx, "correct": c})
        except Exception:
            traceback.print_exc(limit=1)
            sweep()
        csv_append(out_csv, F2, rows)
    n2 = max(st2["n"], 1)
    res["slm1_acc"] = st2["ok"] / n2
    res["slm1_n"] = st2["n"]
    res["slm1_strategies"] = dict(strat2.most_common(4))

    lat = []
    for i in range(min(N_LAT, N)):
        t1 = time.time()
        gen_safe(s1, s1tok, [PROMPT_1SLM.replace("{question}", QS[i])], SLM1_MAX_NEW)
        lat.append(time.time() - t1)
    res["slm1_lat"] = sum(lat) / len(lat)
    print(f"[{LABEL}] 1-SLM: {st2['ok']}/{n2} = {res['slm1_acc']:.2%} "
          f"| {res['slm1_lat']:.2f}s | {res['slm1_strategies']}")

    del s1, s1tok
    sweep()
    vram("1-SLM released")

    ALL[LABEL] = res
    json.dump(ALL, open(f"{OUT_DIR}/ood_summary.json", "w"), indent=2, default=str)

# =============================================================================
# 3.4  OOD TABLE
# =============================================================================
GSM8K_REF = {"general_baseline": 0.4550, "general_guided": 0.5600,
             "general_p": mcnemar(310, 172), "slm1_acc": 0.6975, "n": 1319}

print("\n" + "=" * 100)
print("OUT-OF-DISTRIBUTION GENERALISATION  (fp16, same planner, same parser)")
print("=" * 100)
print(f"{'dataset':<18}{'n':>6}{'gen base':>10}{'gen guided':>12}{'delta':>9}"
      f"{'p':>10}{'1-SLM':>9}{'1-SLM lat':>11}")
print("-" * 100)
print(f"{'GSM8K (in-dist)':<18}{GSM8K_REF['n']:>6}{GSM8K_REF['general_baseline']:>9.1%}"
      f"{GSM8K_REF['general_guided']:>11.1%}"
      f"{GSM8K_REF['general_guided']-GSM8K_REF['general_baseline']:>+9.1%}"
      f"{GSM8K_REF['general_p']:>10.4f}{GSM8K_REF['slm1_acc']:>8.1%}{'-':>11}")
for lab, r in ALL.items():
    print(f"{lab+' (OOD)':<18}{r['n']:>6}{r.get('general_baseline',0):>9.1%}"
          f"{r.get('general_guided',0):>11.1%}{r.get('general_delta',0):>+9.1%}"
          f"{r.get('general_p',1):>10.4f}{r.get('slm1_acc',0):>8.1%}"
          f"{r.get('slm1_lat',0):>10.2f}s")
print("-" * 100)
for lab, r in ALL.items():
    if "plan_format_ok" in r:
        print(f"  {lab} plan quality OOD: format {r['plan_format_ok']:.1%} | "
              f"ops-in-vocab {r['plan_ops_valid']:.1%} | "
              f"arithmetic leak {r.get('plan_has_math',0):.1%}")

print(f"\nsaved -> {OUT_DIR}/ood_summary.json and ood_*.csv")
print("""
READING IT
  Guidance delta holds OOD -> the interaction effect is a property of the
    method, not of GSM8K. Strongest version of your claim.
  Delta shrinks or vanishes -> part of the +10.5% was GSM8K-specific. Report it;
    a scoped claim that survives beats a broad one that does not.
  1-SLM drops sharply OOD -> compressed reasoning memorised GSM8K templates.
    A real limitation, and worth a paragraph rather than a footnote.
  Plan format/ops near 100% OOD is direct evidence the planner learned a
    transferable output structure, not dataset-specific phrasing.

DOWNLOAD /kaggle/working BEFORE THE SESSION ENDS.
""")
```
